```python
price_columns = [
    "open",
    "high",
    "low",
    "close",
    "next_weeks_open",
    "next_weeks_close"
]

percentage_columns = [
    "percent_change_price",
    "percent_change_volume_over_last_wk",
    "previous_weeks_volume",
    "percent_change_next_weeks_price"
]
```

```python
print(df.columns.tolist())
```

```python
for col in price_columns:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace("$", "", regex=False)
            .str.replace(",", "", regex=False)
        )
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
for col in percentage_columns:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .str.replace("%", "", regex=False)
            .str.replace(",", "", regex=False)
        )
        df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
# Convert date from text to datetime
df["date"] = pd.to_datetime(df["date"])

# Sort data chronologically
df = df.sort_values(["date", "stock"]).reset_index(drop=True)

# Check the result
print("Date type:", df["date"].dtype)
print("First date:", df["date"].min())
print("Last date:", df["date"].max())

display(df.head())
```

```python
# Columns that contain future information
future_columns = [
    "next_weeks_open",
    "next_weeks_close",
    "percent_change_next_weeks_price"
]

# Create ML dataset without future-information columns
ml_df = df.drop(columns=future_columns)

print("Original shape:", df.shape)
print("ML dataset shape:", ml_df.shape)

print("\nRemaining columns:")
print(ml_df.columns.tolist())
```

```python
# Separate features and target

X = ml_df.drop(columns=["Target"])
y = ml_df["Target"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeature columns:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
# Remove date from ML features
X = X.drop(columns=["date"])

# Identify categorical and numerical columns
categorical_features = ["stock"]

numerical_features = [
    "quarter",
    "open",
    "high",
    "low",
    "close",
    "volume",
    "percent_change_price",
    "percent_change_volume_over_last_wk",
    "previous_weeks_volume",
    "days_to_next_dividend",
    "percent_return_next_dividend"
]

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)

print("\nX shape:", X.shape)
```

```python
# Get the original dates for splitting
dates = df["date"]

# Find the date at the 80% point
unique_dates = sorted(dates.unique())
split_index = int(len(unique_dates) * 0.8)
split_date = unique_dates[split_index]

# Create train/test masks
train_mask = dates < split_date
test_mask = dates >= split_date

# Split features and target
X_train = X[train_mask]
X_test = X[test_mask]

y_train = y[train_mask]
y_test = y[test_mask]

print("Split date:", split_date)

print("\nTraining data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Preprocessing for numerical and categorical features
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

# Fit only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the same preprocessing
X_test_processed = preprocessor.transform(X_test)

print("Original training shape:", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original testing shape:", X_test.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

# Create Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
logistic_model.fit(X_train_processed, y_train)

# Predict on test data
y_pred_logistic = logistic_model.predict(X_test_processed)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred_logistic)
precision = precision_score(y_test, y_pred_logistic)
recall = recall_score(y_test, y_pred_logistic)
f1 = f1_score(y_test, y_pred_logistic)

print("LOGISTIC REGRESSION RESULTS")
print("===========================")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_logistic))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_logistic))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=10,
    random_state=42,
    class_weight="balanced"
)

# Train the model
rf_model.fit(X_train_processed, y_train)

# Make predictions
y_pred_rf = rf_model.predict(X_test_processed)

# Calculate evaluation metrics
rf_accuracy = accuracy_score(y_test, y_pred_rf)
rf_precision = precision_score(y_test, y_pred_rf)
rf_recall = recall_score(y_test, y_pred_rf)
rf_f1 = f1_score(y_test, y_pred_rf)

print("RANDOM FOREST RESULTS")
print("=====================")
print(f"Accuracy  : {rf_accuracy:.4f}")
print(f"Precision : {rf_precision:.4f}")
print(f"Recall    : {rf_recall:.4f}")
print(f"F1 Score  : {rf_f1:.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_rf))
```

```python
# Compare Logistic Regression and Random Forest

comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest"
    ],
    "Accuracy": [
        accuracy,
        rf_accuracy
    ],
    "Precision": [
        precision,
        rf_precision
    ],
    "Recall": [
        recall,
        rf_recall
    ],
    "F1 Score": [
        f1,
        rf_f1
    ]
})

# Display results
display(comparison.round(4))
```

```python
# Create trading signals from Random Forest predictions

test_results = df.loc[test_mask].copy()

# Add model predictions
test_results["Predicted_Target"] = y_pred_rf

# Convert predictions into trading signals
# 1 = Buy/Hold
# 0 = No position
test_results["Signal"] = test_results["Predicted_Target"]

print("Trading signals:")
print(test_results["Signal"].value_counts())

print("\nSample results:")
display(
    test_results[
        ["date", "stock", "close", "percent_change_next_weeks_price",
         "Target", "Predicted_Target", "Signal"]
    ].head(10)
)
```

```python
# Calculate strategy return
# If Signal = 1, take the actual next-week return.
# If Signal = 0, stay out of the market.

test_results["Strategy_Return"] = (
    test_results["Signal"]
    * test_results["percent_change_next_weeks_price"]
)

# Convert percentage return to decimal
test_results["Strategy_Return_Decimal"] = (
    test_results["Strategy_Return"] / 100
)

print("Strategy return statistics:")
print(test_results["Strategy_Return"].describe())

print("\nFirst 10 strategy returns:")
display(
    test_results[
        ["date", "stock", "Signal",
         "percent_change_next_weeks_price",
         "Strategy_Return"]
    ].head(10)
)
```

```python
# Calculate equal-weighted portfolio return for each week
weekly_returns = (
    test_results
    .groupby("date")["Strategy_Return_Decimal"]
    .mean()
    .sort_index()
)

# Initial portfolio value
initial_capital = 100000

# Calculate cumulative portfolio value
portfolio_value = initial_capital * (1 + weekly_returns).cumprod()

# Final portfolio value
final_capital = portfolio_value.iloc[-1]

# Total return
total_return = ((final_capital - initial_capital) / initial_capital) * 100

print("PORTFOLIO PERFORMANCE")
print("=====================")
print(f"Initial Capital : ₹{initial_capital:,.2f}")
print(f"Final Capital   : ₹{final_capital:,.2f}")
print(f"Total Return    : {total_return:.2f}%")

print("\nWeekly portfolio returns:")
display(weekly_returns.to_frame("Weekly_Return").head(10))
```

```python
import numpy as np

# -----------------------------
# Maximum Drawdown
# -----------------------------

# Calculate running maximum portfolio value
running_max = portfolio_value.cummax()

# Calculate drawdown
drawdown = (portfolio_value - running_max) / running_max

# Maximum drawdown
max_drawdown = drawdown.min() * 100


# -----------------------------
# Sharpe Ratio
# -----------------------------

# Weekly Sharpe ratio
if weekly_returns.std() != 0:
    sharpe_ratio = (
        weekly_returns.mean() / weekly_returns.std()
    ) * np.sqrt(52)
else:
    sharpe_ratio = 0


# -----------------------------
# Number of Trades
# -----------------------------

number_of_trades = int(test_results["Signal"].sum())


# -----------------------------
# Display results
# -----------------------------

print("RISK AND TRADING METRICS")
print("========================")
print(f"Maximum Drawdown : {max_drawdown:.2f}%")
print(f"Sharpe Ratio     : {sharpe_ratio:.4f}")
print(f"Number of Trades : {number_of_trades}")
```

```python
# Create portfolio history table

portfolio_history = pd.DataFrame({
    "Date": portfolio_value.index,
    "Weekly_Return": weekly_returns.values,
    "Portfolio_Value": portfolio_value.values
})

print("Portfolio History:")
display(portfolio_history)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.plot(
    portfolio_history["Date"],
    portfolio_history["Portfolio_Value"],
    marker="o"
)

plt.title("ML Trading Strategy Portfolio Performance")
plt.xlabel("Date")
plt.ylabel("Portfolio Value (₹)")
plt.grid(True)

plt.show()
```
